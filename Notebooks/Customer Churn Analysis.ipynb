```python
#churn rate by satisfaction score group
churn_satisfaction = pd.merge(
    support_tickets[["Customer_ID", "Satisfaction_Score"]],
    customers[["Customer_ID", "Customer_Status"]],
    on="Customer_ID",
    how="left"
)

churn_satisfaction["Satisfaction_Group"] = pd.cut(
    churn_satisfaction["Satisfaction_Score"],
    bins=[0, 2, 3, 5],
    labels=["Low", "Medium", "High"]
)
churn_rate = (
    churn_satisfaction.groupby("Satisfaction_Group")["Customer_Status"]
    .apply(lambda x: (x == "Churned").mean() * 100)
)

print(churn_rate)
```

```python
usage_issue_churn = pd.merge(
    monthly_usage[["Customer_ID", "Usage_Issue_Count"]],
    customers[["Customer_ID", "Customer_Status"]],
    on="Customer_ID",
    how="left"
)

avg_usage_issues = (
    usage_issue_churn.groupby("Customer_Status")["Usage_Issue_Count"]
    .mean()
)

print(avg_usage_issues)
```

```python
avg_usage_issues = (
    usage_issue_churn.groupby("Customer_Status")["Usage_Issue_Count"]
    .mean()
)

difference = (
    avg_usage_issues["Churned"] - avg_usage_issues["Active"]
)

percentage_difference = (
    difference / avg_usage_issues["Active"]
) * 100

print("Average Usage Issues:")
print(avg_usage_issues.round(2))

print("\nDifference:", round(difference, 2))
print("Percentage Difference:", round(percentage_difference, 2), "%")
```

```python
# Filter only churned customers
churned_customers = customers[
    customers["Customer_Status"] == "Churned"
].copy()

# 1. Age profile
age_profile = churned_customers["Age"].agg(
    ["mean", "median", "min", "max"]
).round(2)

# 2. Gender profile
gender_profile = (
    churned_customers["Gender"]
    .value_counts()
)

# 3. Region profile
region_profile = (
    churned_customers["Region"]
    .value_counts()
)

# 4. Acquisition Channel profile
acquisition_profile = (
    churned_customers["Acquisition_Channel"]
    .value_counts()
)

# 5. Primary Device profile
device_profile = (
    churned_customers["Primary_Device"]
    .value_counts()
)

# 6. Top 10 cities by churn
city_profile = (
    churned_customers["City"]
    .value_counts()
    .head(10)
)

# 7. Top 10 states by churn
state_profile = (
    churned_customers["State"]
    .value_counts()
    .head(10)
)

# Display results
print("AGE PROFILE")
print(age_profile)

print("\nGENDER PROFILE")
print(gender_profile)

print("\nREGION PROFILE")
print(region_profile)

print("\nACQUISITION CHANNEL PROFILE")
print(acquisition_profile)

print("\nPRIMARY DEVICE PROFILE")
print(device_profile)

print("\nTOP 10 CITIES BY CHURN")
print(city_profile)

print("\nTOP 10 STATES BY CHURN")
print(state_profile)
```
